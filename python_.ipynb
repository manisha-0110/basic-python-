```python
x = "AAACCDDDAAD"
res = ""
c = 1

for i in range(len(x)):
    if i + 1 < len(x) and x[i] == x[i + 1]:
        c += 1
    else:
        res += x[i] + str(c)
        c = 1

print(res)
```

```python
def manisha(a):
    if a == 1:
        return 1
    else:
        return a * manisha(a - 1)

a = 6
print(manisha(a))
```

```python
def flower(a):
    if a == 4:
        return
    print(a + 10, end=" ")
    flower(a + 1)
    print(a, end=" ")

flower(1)
```

```python
def raju(a):
    if a == 1:
        return
    a -= 1
    raju(a)
    print("hai")
    raju(a)

raju(5)
```

```python
def raju(a):
    if a == 5:
        return 1

    B = raju(a + 1)
    print("hai")
    return B + 2

res = raju(1)
print(res)
```

```python
class abc:
    num = 100

    def set_dim(self, a, b):
        self.total = a + b + abc.num

    def display(self):
        print("Addition is", self.total)

a = abc()
b = abc()

a.set_dim(20, 30)
b.set_dim(30, 50)

a.display()
b.display()
```

```python
class abc:
    num = 100

    def set_dim(self, a, b):
        self.total = a + b + abc.num
        self.display()

    def display(self):
        print("Addition is", self.total)

a = abc()
a.set_dim(20, 30)
a.set_dim(30, 50)
```

```python
def Manisha(a):
    if a == 1:
        return
    i = 2
    while i <= a:
        if a % i == 0:
            print(i, end=" ")
            Manisha(a // i)
            break
        i += 1

a = int(input("Enter any number: "))
Manisha(a)
```

```python
class Rect:
    def __init__(self, a, b):
        self.area = a * b

    def display(self):
        print("Area is", self.area)

r = Rect(20, 30)
r.display()
```

```python
class circle:
    def __init__(self):
        self.r = int(input("Enter radius: "))

    def display(self):
        print("Area is", 3.14 * self.r ** 2)

n = int(input("Enter number of circles: "))
l = []

for i in range(n):
    c = circle()
    l.append(c)
    c.display()
    print("__________")
```
