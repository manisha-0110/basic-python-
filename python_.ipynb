```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        s = "".join(i.lower() for i in s if i.isalnum())
        return s == s[::-1]
```

```python
class Solution:
    def isPerfectSquare(self, num: int) -> bool:
        x = int(num ** 0.5)
        return x * x == num
```

```python
class Solution:
    def runningSum(self, nums):
        for i in range(1, len(nums)):
            nums[i] = nums[i] + nums[i - 1]
        return nums


# Test
obj = Solution()
print(obj.runningSum([1, 2, 3, 4]))
```

```python
n = 11

a = [[' ' for _ in range(n)] for _ in range(n)]

for i in range(n):
    for j in range(n):
        if j == 0:
            a[i][j] = '*'
        elif j == n - 1:
            a[i][j] = '*'
        elif i == j and i <= n // 2:
            a[i][j] = '*'
        elif i + j == n - 1 and i <= n // 2:
            a[i][j] = '*'

for i in range(n):
    for j in range(n):
        print(a[i][j], end=' ')
    print()
```

```python
def sq():
    s = int(input("Enter side: "))
    print("Area of square:", s * s)

def rect(l, b):
    print("Area of rectangle:", l * b)

def cir():
    r = int(input("Enter radius: "))
    print("Area of circle:", 3.14 * r * r)

def tri(b, h):
    print("Area of triangle:", 0.5 * b * h)

while True:
    print("\n1. SQ")
    print("2. RECT")
    print("3. CIR")
    print("4. TRI")
    print("5. EXIT")

    ch = int(input("Enter your choice: "))

    if ch == 1:
        sq()

    elif ch == 2:
        l = int(input("Enter length: "))
        b = int(input("Enter breadth: "))
        rect(l, b)

    elif ch == 3:
        cir()

    elif ch == 4:
        b = int(input("Enter base: "))
        h = int(input("Enter height: "))
        tri(b, h)

    elif ch == 5:
        print("Exiting...")
        break

    else:
        print("Invalid choice!")
```

```python
def prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def sod(n):
    return sum(int(i) for i in str(n))

def every(n):
    return all(prime(int(d)) for d in str(n) if d != '0')

for i in range(100, 1000):
    if prime(i) and prime(sod(i)) and every(i):
        print(i, end=" ")
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        newnode = Node(data)

        if self.head is None:
            self.head = newnode
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = newnode

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")

l = LinkedList()
l.insert(100)
l.insert(200)
l.insert(300)
l.insert(400)
l.insert(500)

l.display()
```
